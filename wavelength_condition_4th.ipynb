ここで、波数kとしたときに、
$$
f(n\Delta x) = \exp{(ikn\Delta x)}
$$
とする。すると、この微分は、
$$
\frac{d}{dx}f(n\Delta x)=ik_e \exp{(ikn\Delta x)}
$$
となります。$k_e$は実効的な波数で、解析的には$k_e=k$となります。一方で、2次精度の中心差分で評価しますと、
$$
\frac{d}{dx}f(n\Delta x) \sim \frac{\exp{{ik(n\Delta x+1/2\Delta x)}}-\exp{{ik(n\Delta x-1/2\Delta x)}}}{\Delta x} \\
=i\frac{\sin(k\Delta x/2)}{\Delta x/2}f(n\Delta x)
$$
となります。2次精度の場合、$k_e$は、
$$
k_e = \frac{\sin(k\Delta x/2)}{\Delta x/2}
$$
となります。

```python
import numpy as np
import matplotlib.pyplot as plt

kdx   = np.arange(0.02,4,0.02)
error = 1-np.sin(kdx/2)/(kdx/2) 
```

$k$と$k_e$の誤差を
$$
\frac{k-k_e}{k}
$$
として相対的に評価するとすると、
$$
\frac{k-k_e}{k}=\frac{k-\frac{\sin(k\Delta x/2)}{\Delta x/2}}{k}=1-\frac{\sin{(k\Delta x/2)}}{k\Delta x/2}
$$
となりますので、$k\Delta x$を変数として図を記します。

```python
plt.figure(figsize=(6, 3))
plt.rcParams.update({'font.size': 10})
plt.rcParams['font.family'] = 'Arial' # font familyの設定
plt.rcParams['xtick.direction'] = 'in' #x軸の目盛りの向き
plt.rcParams['ytick.direction'] = 'in' #y軸の目盛りの向き
plt.rcParams["xtick.minor.visible"] = True  #x軸補助目盛りの追加
plt.rcParams["ytick.minor.visible"] = True  #y軸補助目盛りの追加

# === プロット ===
plt.rcParams.update({'font.size': 10})
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams["xtick.minor.visible"] = True  
plt.rcParams["ytick.minor.visible"] = True 

plt.plot(kdx,error)
plt.xlabel(r"$k\Delta x$",fontsize=14)
plt.ylabel(r"Error $(k-k_e)/k$",fontsize=14)
plt.xlim(0,3)
plt.ylim(0,0.1)
plt.show()
```

$k\Delta x = 0.4$程度まで誤差が1 %以下となりそうだが、
$$
k\Delta x=\frac{2\pi \Delta x}{\lambda}
$$
となるので、
$$
\frac{\lambda}{\Delta x} = \frac{2\pi}{0.4} \sim 15
$$
逆数をとれば波長あたりの格子数となり、15格子程度必要なこととなる。

## 4次精度のスタッガード格子差分の場合

4次精度のスタッガード格子差分では、$f$ の微分を半格子ずれた点で、
$$
\frac{d}{dx}f(n\Delta x) \sim \frac{1}{\Delta x}\left[\frac{9}{8}\left\{f\left(n\Delta x+\tfrac{1}{2}\Delta x\right)-f\left(n\Delta x-\tfrac{1}{2}\Delta x\right)\right\}-\frac{1}{24}\left\{f\left(n\Delta x+\tfrac{3}{2}\Delta x\right)-f\left(n\Delta x-\tfrac{3}{2}\Delta x\right)\right\}\right]
$$
と評価します。$f(n\Delta x) = \exp{(ikn\Delta x)}$ を代入すると、
$$
\frac{d}{dx}f(n\Delta x) = i\,\frac{1}{\Delta x}\left\{\frac{9}{4}\sin\left(\frac{k\Delta x}{2}\right)-\frac{1}{12}\sin\left(\frac{3k\Delta x}{2}\right)\right\}f(n\Delta x)
$$
となり、4次精度の場合の $k_e$ は、
$$
k_e = \frac{1}{\Delta x}\left\{\frac{9}{4}\sin\left(\frac{k\Delta x}{2}\right)-\frac{1}{12}\sin\left(\frac{3k\Delta x}{2}\right)\right\}
$$
となります。相対誤差は、
$$
\frac{k-k_e}{k}=1-\frac{\frac{9}{4}\sin\left(\frac{k\Delta x}{2}\right)-\frac{1}{12}\sin\left(\frac{3k\Delta x}{2}\right)}{k\Delta x}
$$
です。2次精度の結果と比べて図に記します。

```python
error2 = 1 - np.sin(kdx/2)/(kdx/2)
error4 = 1 - (9/4*np.sin(kdx/2) - 1/12*np.sin(3*kdx/2))/kdx

plt.figure(figsize=(6, 3))
plt.plot(kdx, error2, label="2nd order")
plt.plot(kdx, error4, label="4th order")
plt.axhline(0.01, color="gray", ls="--", lw=0.8)  # 1 % level
plt.xlabel(r"$k\Delta x$", fontsize=14)
plt.ylabel(r"Error $(k-k_e)/k$", fontsize=14)
plt.xlim(0, 3)
plt.ylim(0, 0.1)
plt.legend()
plt.show()

# 誤差が 1 % を超える最初の k*dx と、必要な 1 波長あたりの格子数
for name, err in [("2nd", error2), ("4th", error4)]:
    k1 = kdx[np.argmax(err > 0.01)]
    print(f"{name} order: k*dx = {k1:.2f}, lambda/dx = {2*np.pi/k1:.1f}")
```

```python
# 横軸を 1 波長あたりの格子点数 lambda/dx = 2*pi/(k*dx) にした図
gpw = 2*np.pi/kdx   # grid points per wavelength

plt.figure(figsize=(6, 3))
plt.plot(gpw, error2, label="2nd order")
plt.plot(gpw, error4, label="4th order")
plt.axhline(0.01, color="gray", ls="--", lw=0.8)  # 1 % level
plt.xlabel(r"Grid points per wavelength $\lambda/\Delta x$", fontsize=14)
plt.ylabel(r"Error $(k-k_e)/k$", fontsize=14)
plt.xlim(2, 30)
plt.ylim(0, 0.1)
plt.legend()
plt.show()
```
